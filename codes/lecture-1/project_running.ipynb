## 3. State the inputs

For this synthetic household, both amounts are **EUR per month**. Run the input cell, then the calculation cell, then the display cell below.


```python
income = 1000
consumption = 800
```

## 4. Calculate saving

Saving is income minus consumption. Predict the answer before running the next two cells.


```python
saving = income - consumption
```

## 5. Display and check

Expected saving: EUR 200 per month. Printing a result and calculating it are separate steps.


```python
print("Monthly saving (EUR):", saving)
```

## 6. Experiment: a result can be stale

1. First obtain `200` by running all three household code cells in order.
2. Change `income` in the input cell from `1000` to `1200` and run **only that input cell**.
3. Skip the saving calculation cell. Run **only the display cell**.
4. Why does the output still say `200`? Which calculation did not run?

The kernel remembers the earlier value of `saving`. Updating `income` alone does not update it.


## 7. Rebuild the result from a clean start

Predict the answer with income `1200` and consumption `800`: saving should be `400`.

Choose **Kernel → Restart Kernel and Run All Cells…** and confirm. This clears remembered variables and runs all the code cells in page order. Check that the path result is `True` and saving is `400`.

Then restore income to `1000`, restart and run all again, check `200`, and save the notebook. Saved output records an earlier run; it is not the same thing as the kernel’s current memory.
