## A complete saving-share function

````python
def saving_share(income, consumption):
    if income < 0 or consumption < 0:
        raise ValueError("Amounts must be non-negative")
    if income == 0:
        return None
    return (income - consumption) / income

print(saving_share(1000, 800))
````

```text
0.2
```

`raise ValueError(...)` stops this call with an explanation. `return` ends the function call immediately.

## Check ordinary, boundary, and negative-saving cases

````python
def saving_share(income, consumption):
    if income < 0 or consumption < 0:
        raise ValueError("Amounts must be non-negative")
    if income == 0:
        return None
    return (income - consumption) / income

print(saving_share(1000, 800))
print(saving_share(1000, 1000))
print(saving_share(1000, 1100))
print(saving_share(0, 100) is None)
````

True
