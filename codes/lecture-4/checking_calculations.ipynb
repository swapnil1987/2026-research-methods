## Compare floats with an explicit tolerance

````python
from math import isclose

total = 0.1 + 0.2
print(total == 0.3)
print(isclose(total, 0.3, rel_tol=1e-9, abs_tol=1e-12))
assert isclose(total, 0.3, rel_tol=1e-9, abs_tol=1e-12)
````

## State the function's contract before testing it

````python
def saving_share(income, consumption):
    if income < 0 or consumption < 0:
        raise ValueError("Amounts must be non-negative")
    if income == 0:
        return None
    return (income - consumption) / income
````

- Inputs are numeric, finite monthly amounts in the same currency.
- Negative income or consumption is outside this teaching specification.
- Zero income returns `None`; consumption above income can produce a negative share.

## Check the contract, including the boundary

Using the function just defined:

````python
from math import isclose

assert isclose(saving_share(1000, 800), 0.2)
assert isclose(saving_share(800, 850), -0.0625)
assert saving_share(0, 100) is None
print("Three checks passed")
````

```text
Three checks passed
```

`saving_share(-100, 50)` should raise `ValueError`, not return a share.
