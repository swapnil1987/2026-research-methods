## Convert deliberately before calculating

````python
income_text = "1000"
price_text = "2.50"

income = int(income_text)
price = float(price_text)

print(income + 200)
print(2 * price)
````

## Format a readable result with an f-string

````python
saving = 200
saving_share = 0.2

print(f"Saving: {saving:.2f} EUR per month")
print(f"Saving rate: {saving_share:.1%}")
````

```text
Saving: 200.00 EUR per month
Saving rate: 20.0%
```

- `f"..."` inserts values inside `{...}`.
- `.2f` displays two decimal places; `.1%` displays a percentage with one decimal place.

## A comparison produces a Boolean

````python
income = 1000
consumption = 800
spends_less_than_income = consumption < income

print(spends_less_than_income)
print(type(spends_less_than_income))
````

```text
True
```

Python's Boolean values are **`True`** and **`False`**, with capital initial letters.

## Compare values explicitly

````python
print(800 < 1000)
print(1000 <= 1000)
print(1000 > 800)
print(1000 >= 1000)
print(200 == 200)
print(200 != 200)
````

Outputs: `True`, `True`, `True`, `True`, `True`, `False`.

`<=` and `>=` include equality. `==` tests equality; `!=` tests inequality.

## Assignment and equality do different jobs

````python
income = 1000          # assign a value
print(income == 1000)  # ask whether it is equal
print(income == 800)
print(income)
````

```text
True
False
1000
```

Testing equality does **not** change the value assigned to `income`.

## Combine Boolean questions

````python
income = 1000
household_size = 2

print((income <= 1200) and (household_size >= 2))
print((income < 900) or (household_size >= 2))
print(not (income <= 1200))
````

```text
True
True
False
```

**`and`:** both. **`or`:** at least one. **`not`:** reverse the Boolean result.

## Indexing starts at zero

````python
incomes = [1000, 1200, 800]

print(incomes[0])
print(incomes[1])
print(incomes[2])
````

```text

```

The first position is `0`, the second is `1`, and the third is `2`.

## Negative indices count from the end

````python
incomes = [1000, 1200, 800]

print(incomes[-1])
print(incomes[-2])
````

```text
800
1200
```

For this three-item list, `incomes[3]` raises `IndexError`: that position does not exist.

## A slice selects several positions

````python
incomes = [1000, 1200, 800, 1500]

print(incomes[0:2])
print(incomes[:2])
print(incomes[1:])
````

```text
[1000, 1200]
[1000, 1200]
[1200, 800, 1500]
```

`start:stop` includes **start** and excludes **stop**. An omitted endpoint uses the corresponding end.

## Change one list item

````python
incomes = [1000, 1200, 800]
incomes[1] = 1250

print(incomes)
````

```text
[1000, 1250, 800]
```

This changes the value at position `1`. The number of items stays the same.

## Append one new item

````python
incomes = [1000, 1200, 800]
incomes.append(1500)

print(incomes)
````

```text
[1000, 1200, 800, 1500]
```

`.append(...)` adds an item at the end and changes the list in place.

Do **not** write `incomes = incomes.append(1500)`; `append` returns `None`.

## Count, total, and average a list

````python
incomes = [1000, 1200, 800]

print(len(incomes))
print(sum(incomes))
print(sum(incomes) / len(incomes))
print(min(incomes), max(incomes))
````

```text
3
3000
1000.0
800 1200
```

These are built-in tools. This average assumes the list is **non-empty** and numeric.

## Check a household total in two steps

````python
incomes = [1000, 1200, 800]
consumptions = [800, 900, 850]

total_saving = sum(incomes) - sum(consumptions)
mean_saving = total_saving / len(incomes)

print(total_saving)
print(f"Mean saving: {mean_saving:.2f} EUR per month")
````

```text
450
Mean saving: 150.00 EUR per month
```

The inputs must cover the **same households and period**, in the same units.

## Copy a list before making an independent change

````python
incomes = [1000, 1200, 800]
alias = incomes
separate = incomes.copy()

alias[0] = 999
print(incomes)
print(separate)
````

```text
[999, 1200, 800]
[1000, 1200, 800]
```

`alias = incomes` gives the **same list** another name. `.copy()` creates a separate list here.
